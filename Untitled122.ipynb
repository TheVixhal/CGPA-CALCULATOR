```python
import numpy as np
import matplotlib.pyplot as plt

# --- Activation Functions and Derivatives ---
# We use ReLU for hidden layers for its simplicity and efficiency,
# and a linear function for the output layer, which is standard for regression tasks.
def relu(x):
    return np.maximum(0, x)

def relu_derivative(x):
    return np.where(x > 0, 1, 0)

def linear(x):
    return x

def linear_derivative(x):
    return np.ones_like(x)

# --- Loss Function ---
# Mean Squared Error is a common choice for regression problems.
def mean_squared_error(y_true, y_pred):
    return np.mean((y_true - y_pred) ** 2)

def mse_derivative(y_true, y_pred):
    return y_pred - y_true

class DynamicPruningNN:
    """
    A neural network with a novel architecture featuring:
    1. Inline Backpropagation: Layer-by-layer updates during the forward pass.
    2. Dynamic Shortcuts: Neurons can bypass layers if they become predictive.
    3. Contribution-Based Pruning: Removes neurons with negative contributions.
    """
    def __init__(self, layer_sizes, learning_rate=0.001):
        self.layer_sizes = layer_sizes
        self.learning_rate = learning_rate
        self.num_layers = len(layer_sizes)

        # --- Initialize Network Parameters ---
        # We use dictionaries to store weights and biases for clarity.
        self.weights = {}
        self.biases = {}
        # We track active neurons for pruning. All start as active.
        self.active_neurons = {}
        # Contribution scores help decide which neurons to prune.
        self.neuron_contributions = {}
        # Stores the dynamic shortcut connections.
        self.shortcuts = {}

        for i in range(1, self.num_layers):
            # Initialize weights with small random values to break symmetry.
            self.weights[i] = np.random.randn(layer_sizes[i-1], layer_sizes[i]) * 0.1
            self.biases[i] = np.zeros((1, layer_sizes[i]))
            # All neurons are active initially.
            self.active_neurons[i] = np.ones(layer_sizes[i], dtype=bool)
            # Contributions start at zero.
            self.neuron_contributions[i] = np.zeros(layer_sizes[i])

    def _temp_predict(self, start_layer_idx, activation_in):
        """
        Performs a temporary forward pass from a given hidden layer to the output.
        This is the core of the "inline" feedback mechanism.
        """
        temp_activation = activation_in
        for i in range(start_layer_idx, self.num_layers - 1):
            # Get the active neuron masks for the previous and current layers in this temp pass
            prev_active_neurons = self.active_neurons.get(i, np.ones(self.weights[i+1].shape[0], dtype=bool))
            curr_active_neurons = self.active_neurons[i+1]

            # Get the sub-matrix of weights connecting only active neurons
            active_w = self.weights[i+1][np.ix_(prev_active_neurons, curr_active_neurons)]
            active_b = self.biases[i+1][:, curr_active_neurons]

            # Filter the input activation from the previous step to match the weight matrix
            filtered_input = temp_activation[:, prev_active_neurons]

            # Calculate the 'z' value for only the active part of the network
            z_active = np.dot(filtered_input, active_w) + active_b

            # Create a full-sized 'z' vector (padded with zeros) to maintain consistent shapes
            z_full = np.zeros((1, self.layer_sizes[i+1]))
            z_full[:, curr_active_neurons] = z_active

            if i < self.num_layers - 2: # Hidden layers
                temp_activation = relu(z_full)
            else: # Output layer
                temp_activation = linear(z_full)
        return temp_activation

    def train(self, X, y, epochs, prune_interval=50, shortcut_threshold=0.1):
        """
        Main training loop for the network.
        """
        loss_history = []
        print("--- Starting Training ---")
        for epoch in range(epochs):
            total_loss = 0
            for x_sample, y_sample in zip(X, y):
                x_sample = x_sample.reshape(1, -1)
                y_sample = y_sample.reshape(1, -1)

                # --- Forward Pass and Dynamic Shortcut Creation ---
                final_pred, activations, zs = self._forward_pass_and_check_shortcuts(x_sample, y_sample, shortcut_threshold, epoch)

                total_loss += mean_squared_error(y_sample, final_pred)

                # --- Unified Backpropagation ---
                self._backpropagate(y_sample, final_pred, activations, zs)

            # --- Periodic Pruning ---
            if (epoch + 1) % prune_interval == 0:
                self._prune_network()

            avg_loss = total_loss / len(X)
            loss_history.append(avg_loss)
            if (epoch + 1) % 10 == 0:
                print(f"Epoch {epoch+1}/{epochs}, Loss: {avg_loss:.6f}")

        print("--- Training Finished ---")
        return loss_history

    def _forward_pass_and_check_shortcuts(self, x_sample, y_sample, shortcut_threshold, epoch):
        """
        Performs a full forward pass while also checking if any neurons
        should be converted into shortcuts.
        """
        activations = {0: x_sample}
        zs = {}
        # Hidden layers
        for i in range(1, self.num_layers - 1):
            prev_active_neurons = self.active_neurons.get(i-1, np.ones(self.weights[i].shape[0], dtype=bool))
            active_w = self.weights[i][np.ix_(prev_active_neurons, self.active_neurons[i])]
            active_b = self.biases[i][:, self.active_neurons[i]]

            z_full = np.zeros((1, self.layer_sizes[i]))
            z_active = np.dot(activations[i-1][:, prev_active_neurons], active_w) + active_b
            z_full[:, self.active_neurons[i]] = z_active

            zs[i] = z_full
            activations[i] = relu(zs[i])

            # --- Inline Check for Shortcut Creation ---
            temp_pred = self._temp_predict(i, activations[i])
            local_error = mse_derivative(y_sample, temp_pred)
            if np.mean(np.abs(local_error)) < shortcut_threshold:
                best_neuron_idx = np.argmax(activations[i])
                if self.active_neurons[i][best_neuron_idx] and (i, best_neuron_idx) not in self.shortcuts:
                    shortcut_weight = np.random.randn(1,1) * 0.1
                    self.shortcuts[(i, best_neuron_idx)] = shortcut_weight
                    print(f"Epoch {epoch}: Created shortcut from Layer {i}, Neuron {best_neuron_idx}")

        # Output layer
        output_layer_idx = self.num_layers - 1
        prev_active_neurons = self.active_neurons[output_layer_idx - 1]

        z_out = np.dot(activations[output_layer_idx-1][:, prev_active_neurons], self.weights[output_layer_idx][prev_active_neurons, :]) + self.biases[output_layer_idx]
        zs[output_layer_idx] = z_out
        activations[output_layer_idx] = linear(zs[output_layer_idx])

        # Add shortcut contributions to the final output
        shortcut_output = 0
        for (layer_idx, neuron_idx), weight in self.shortcuts.items():
            neuron_activation = activations[layer_idx][0, neuron_idx]
            shortcut_output += neuron_activation * weight

        final_output = activations[output_layer_idx] + shortcut_output
        return final_output, activations, zs

    def _backpropagate(self, y, y_pred, activations, zs):
        """
        Performs a unified backpropagation pass that updates all main path weights,
        shortcut weights, and contribution scores.
        """
        deltas = {}

        # Start with the error at the output layer
        output_idx = self.num_layers - 1
        error_signal = mse_derivative(y, y_pred)
        deltas[output_idx] = error_signal * linear_derivative(zs[output_idx])

        # --- Update Shortcut Weights ---
        for (layer_idx, neuron_idx), weight in self.shortcuts.items():
            neuron_activation = activations[layer_idx][0, neuron_idx]
            # Gradient for shortcut is error * neuron's activation
            d_shortcut_weight = error_signal * neuron_activation
            self.shortcuts[(layer_idx, neuron_idx)] -= self.learning_rate * d_shortcut_weight

        # Propagate error backwards through hidden layers
        for i in range(output_idx - 1, 0, -1):
            error = np.dot(deltas[i+1], self.weights[i+1].T)
            delta_full = np.zeros((1, self.layer_sizes[i]))
            active_error = error[:, self.active_neurons[i]]
            active_z = zs[i][:, self.active_neurons[i]]
            delta_active = active_error * relu_derivative(active_z)
            delta_full[:, self.active_neurons[i]] = delta_active
            deltas[i] = delta_full

        # Update main path weights, biases, and contribution scores
        for i in range(1, self.num_layers):
            prev_active_mask = self.active_neurons.get(i-1, np.ones(self.weights[i].shape[0], dtype=bool))
            curr_active_mask = self.active_neurons[i]

            d_weights = np.dot(activations[i-1][:, prev_active_mask].T, deltas[i][:, curr_active_mask])

            # --- Contribution Tracking ---
            # Add a small decay to all scores
            self.neuron_contributions[i] *= 0.999
            # Add the current impact to the neurons that were just updated
            if d_weights.shape[1] > 0: # Ensure there are active neurons to update
                neuron_impact = np.mean(np.abs(d_weights), axis=0)
                self.neuron_contributions[i][curr_active_mask] += neuron_impact

            # Update weights and biases
            update_matrix = np.zeros_like(self.weights[i])
            ix_ = np.ix_(prev_active_mask, curr_active_mask)
            update_matrix[ix_] = d_weights
            self.weights[i] -= self.learning_rate * update_matrix
            self.biases[i][:, curr_active_mask] -= self.learning_rate * np.sum(deltas[i][:, curr_active_mask], axis=0)

    def _prune_network(self, prune_threshold=0.0):
        """Removes neurons with a contribution score below the threshold."""
        print("\n--- Checking for Pruning ---")
        for i in range(1, self.num_layers - 1): # Only prune hidden layers
            for neuron_idx in range(self.layer_sizes[i]):
                # Check if neuron is active and its score is below threshold
                if self.active_neurons[i][neuron_idx] and self.neuron_contributions[i][neuron_idx] < prune_threshold:
                    self.active_neurons[i][neuron_idx] = False
                    print(f"Pruned Neuron {neuron_idx} in Layer {i}. Score: {self.neuron_contributions[i][neuron_idx]:.4f}")
        print("--------------------------\n")

    def predict(self, X):
        """Make predictions on new data with the trained network."""
        predictions = []
        for x_sample in X:
            x_sample = x_sample.reshape(1, -1)
            # Use a simplified forward pass for prediction that respects pruning
            activations = {0: x_sample}
            for i in range(1, self.num_layers - 1):
                prev_active_neurons = self.active_neurons.get(i-1, np.ones(self.weights[i].shape[0], dtype=bool))
                active_w = self.weights[i][np.ix_(prev_active_neurons, self.active_neurons[i])]
                active_b = self.biases[i][:, self.active_neurons[i]]

                z_full = np.zeros((1, self.layer_sizes[i]))
                z_active = np.dot(activations[i-1][:, prev_active_neurons], active_w) + active_b
                z_full[:, self.active_neurons[i]] = z_active
                activations[i] = relu(z_full)

            output_layer_idx = self.num_layers - 1
            prev_active_neurons = self.active_neurons[output_layer_idx - 1]
            z_out = np.dot(activations[output_layer_idx-1][:, prev_active_neurons], self.weights[output_layer_idx][prev_active_neurons, :]) + self.biases[output_layer_idx]
            pred = linear(z_out)

            shortcut_output = 0
            for (layer_idx, neuron_idx), weight in self.shortcuts.items():
                neuron_activation = activations[layer_idx][0, neuron_idx]
                shortcut_output += neuron_activation * weight

            final_pred = pred + shortcut_output
            predictions.append(final_pred.flatten()[0])
        return np.array(predictions)

# --- Main Execution ---
if __name__ == '__main__':
    # 1. Create a simple synthetic dataset
    # Input: [Area (sq ft), Distance from School (km)] -> Output: [Price ($k)]
    np.random.seed(42)
    X_raw = np.random.rand(200, 2)
    X_raw[:, 0] = X_raw[:, 0] * 2000 + 500    # Area between 500-2500 sq ft
    X_raw[:, 1] = X_raw[:, 1] * 10 + 1       # Distance between 1-11 km

    # Price is a function of Area (positive) and Distance (negative) + noise
    y_raw = (X_raw[:, 0] * 0.2) + (X_raw[:, 1] * -5) + np.random.randn(200) * 10 + 50

    # 2. Normalize the data (crucial for training)
    X_mean, X_std = X_raw.mean(axis=0), X_raw.std(axis=0)
    y_mean, y_std = y_raw.mean(), y_raw.std()

    X = (X_raw - X_mean) / X_std
    y = (y_raw - y_mean) / y_std

    # 3. Define and train the network
    # Architecture: 2 inputs -> 10 neurons -> 8 neurons -> 5 neurons -> 1 output
    layer_config = [X.shape[1], 10, 8, 5, 1]
    nn = DynamicPruningNN(layer_sizes=layer_config, learning_rate=0.001)

    epochs = 200
    loss_history = nn.train(X, y, epochs=epochs, prune_interval=50, shortcut_threshold=0.05)

    # 4. Plot the training loss
    plt.figure(figsize=(10, 6))
    plt.plot(range(epochs), loss_history)
    plt.title('Training Loss Over Time')
    plt.xlabel('Epoch')
    plt.ylabel('Mean Squared Error')
    plt.grid(True)
    plt.show()

    # 5. Make predictions and compare
    predictions_normalized = nn.predict(X)
    # De-normalize predictions to see them in the original scale
    predictions = predictions_normalized * y_std + y_mean

    plt.figure(figsize=(10, 6))
    plt.scatter(y_raw, predictions, alpha=0.7, edgecolors='k')
    plt.plot([y_raw.min(), y_raw.max()], [y_raw.min(), y_raw.max()], 'r--', lw=2)
    plt.title('Actual vs. Predicted House Prices')
    plt.xlabel('Actual Price ($k)')
    plt.ylabel('Predicted Price ($k)')
    plt.grid(True)
    plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import fetch_california_housing
from sklearn.model_selection import train_test_split

# --- Activation Functions and Derivatives ---
# We use ReLU for hidden layers for its simplicity and efficiency,
# and a linear function for the output layer, which is standard for regression tasks.
def relu(x):
    return np.maximum(0, x)

def relu_derivative(x):
    return np.where(x > 0, 1, 0)

def linear(x):
    return x

def linear_derivative(x):
    return np.ones_like(x)

# --- Loss Function ---
# Mean Squared Error is a common choice for regression problems.
def mean_squared_error(y_true, y_pred):
    return np.mean((y_true - y_pred) ** 2)

def mse_derivative(y_true, y_pred):
    return y_pred - y_true

class DynamicPruningNN:
    """
    A neural network with a novel architecture featuring:
    1. Inline Backpropagation: Layer-by-layer updates during the forward pass.
    2. Dynamic Shortcuts: Neurons can bypass layers if they become predictive.
    3. Contribution-Based Pruning: Removes neurons with negative contributions.
    """
    def __init__(self, layer_sizes, learning_rate=0.001):
        self.layer_sizes = layer_sizes
        self.learning_rate = learning_rate
        self.num_layers = len(layer_sizes)

        # --- Initialize Network Parameters ---
        # We use dictionaries to store weights and biases for clarity.
        self.weights = {}
        self.biases = {}
        # We track active neurons for pruning. All start as active.
        self.active_neurons = {}
        # Contribution scores help decide which neurons to prune.
        self.neuron_contributions = {}
        # Stores the dynamic shortcut connections.
        self.shortcuts = {}

        for i in range(1, self.num_layers):
            # Initialize weights with small random values to break symmetry.
            self.weights[i] = np.random.randn(layer_sizes[i-1], layer_sizes[i]) * 0.1
            self.biases[i] = np.zeros((1, layer_sizes[i]))
            # All neurons are active initially.
            self.active_neurons[i] = np.ones(layer_sizes[i], dtype=bool)
            # Contributions start at zero.
            self.neuron_contributions[i] = np.zeros(layer_sizes[i])

    def _temp_predict(self, start_layer_idx, activation_in):
        """
        Performs a temporary forward pass from a given hidden layer to the output.
        This is the core of the "inline" feedback mechanism.
        """
        temp_activation = activation_in
        for i in range(start_layer_idx, self.num_layers - 1):
            # Get the active neuron masks for the previous and current layers in this temp pass
            prev_active_neurons = self.active_neurons.get(i, np.ones(self.weights[i+1].shape[0], dtype=bool))
            curr_active_neurons = self.active_neurons[i+1]

            # Get the sub-matrix of weights connecting only active neurons
            active_w = self.weights[i+1][np.ix_(prev_active_neurons, curr_active_neurons)]
            active_b = self.biases[i+1][:, curr_active_neurons]

            # Filter the input activation from the previous step to match the weight matrix
            filtered_input = temp_activation[:, prev_active_neurons]

            # Calculate the 'z' value for only the active part of the network
            z_active = np.dot(filtered_input, active_w) + active_b

            # Create a full-sized 'z' vector (padded with zeros) to maintain consistent shapes
            z_full = np.zeros((1, self.layer_sizes[i+1]))
            z_full[:, curr_active_neurons] = z_active

            if i < self.num_layers - 2: # Hidden layers
                temp_activation = relu(z_full)
            else: # Output layer
                temp_activation = linear(z_full)
        return temp_activation

    def train(self, X, y, epochs, prune_interval=50, shortcut_threshold=0.1):
        """
        Main training loop for the network.
        """
        loss_history = []
        print("--- Starting Training ---")
        for epoch in range(epochs):
            total_loss = 0
            # Shuffle data each epoch
            permutation = np.random.permutation(X.shape[0])
            X_shuffled = X[permutation]
            y_shuffled = y[permutation]

            for x_sample, y_sample in zip(X_shuffled, y_shuffled):
                x_sample = x_sample.reshape(1, -1)
                y_sample = y_sample.reshape(1, -1)

                # --- Forward Pass and Dynamic Shortcut Creation ---
                final_pred, activations, zs = self._forward_pass_and_check_shortcuts(x_sample, y_sample, shortcut_threshold, epoch)

                total_loss += mean_squared_error(y_sample, final_pred)

                # --- Unified Backpropagation ---
                self._backpropagate(y_sample, final_pred, activations, zs)

            # --- Periodic Pruning ---
            if (epoch + 1) % prune_interval == 0:
                self._prune_network()

            avg_loss = total_loss / len(X)
            loss_history.append(avg_loss)
            if (epoch + 1) % 10 == 0:
                print(f"Epoch {epoch+1}/{epochs}, Loss: {avg_loss:.6f}")

        print("--- Training Finished ---")
        return loss_history

    def _forward_pass_and_check_shortcuts(self, x_sample, y_sample, shortcut_threshold, epoch):
        """
        Performs a full forward pass while also checking if any neurons
        should be converted into shortcuts.
        """
        activations = {0: x_sample}
        zs = {}
        # Hidden layers
        for i in range(1, self.num_layers - 1):
            prev_active_neurons = self.active_neurons.get(i-1, np.ones(self.weights[i].shape[0], dtype=bool))
            active_w = self.weights[i][np.ix_(prev_active_neurons, self.active_neurons[i])]
            active_b = self.biases[i][:, self.active_neurons[i]]

            z_full = np.zeros((1, self.layer_sizes[i]))
            z_active = np.dot(activations[i-1][:, prev_active_neurons], active_w) + active_b
            z_full[:, self.active_neurons[i]] = z_active

            zs[i] = z_full
            activations[i] = relu(zs[i])

            # --- Inline Check for Shortcut Creation ---
            temp_pred = self._temp_predict(i, activations[i])
            local_error = mse_derivative(y_sample, temp_pred)
            if np.mean(np.abs(local_error)) < shortcut_threshold:
                best_neuron_idx = np.argmax(activations[i])
                if self.active_neurons[i][best_neuron_idx] and (i, best_neuron_idx) not in self.shortcuts:
                    shortcut_weight = np.random.randn(1,1) * 0.1
                    self.shortcuts[(i, best_neuron_idx)] = shortcut_weight
                    # print(f"Epoch {epoch}: Created shortcut from Layer {i}, Neuron {best_neuron_idx}") # Optional: uncomment for verbose logging

        # Output layer
        output_layer_idx = self.num_layers - 1
        prev_active_neurons = self.active_neurons[output_layer_idx - 1]

        z_out = np.dot(activations[output_layer_idx-1][:, prev_active_neurons], self.weights[output_layer_idx][prev_active_neurons, :]) + self.biases[output_layer_idx]
        zs[output_layer_idx] = z_out
        activations[output_layer_idx] = linear(zs[output_layer_idx])

        # Add shortcut contributions to the final output
        shortcut_output = 0
        for (layer_idx, neuron_idx), weight in self.shortcuts.items():
            neuron_activation = activations[layer_idx][0, neuron_idx]
            shortcut_output += neuron_activation * weight

        final_output = activations[output_layer_idx] + shortcut_output
        return final_output, activations, zs

    def _backpropagate(self, y, y_pred, activations, zs):
        """
        Performs a unified backpropagation pass that updates all main path weights,
        shortcut weights, and contribution scores.
        """
        deltas = {}

        # Start with the error at the output layer
        output_idx = self.num_layers - 1
        error_signal = mse_derivative(y, y_pred)
        deltas[output_idx] = error_signal * linear_derivative(zs[output_idx])

        # --- Update Shortcut Weights ---
        for (layer_idx, neuron_idx), weight in self.shortcuts.items():
            neuron_activation = activations[layer_idx][0, neuron_idx]
            # Gradient for shortcut is error * neuron's activation
            d_shortcut_weight = error_signal * neuron_activation
            self.shortcuts[(layer_idx, neuron_idx)] -= self.learning_rate * d_shortcut_weight

        # Propagate error backwards through hidden layers
        for i in range(output_idx - 1, 0, -1):
            error = np.dot(deltas[i+1], self.weights[i+1].T)
            delta_full = np.zeros((1, self.layer_sizes[i]))
            active_error = error[:, self.active_neurons[i]]
            active_z = zs[i][:, self.active_neurons[i]]
            delta_active = active_error * relu_derivative(active_z)
            delta_full[:, self.active_neurons[i]] = delta_active
            deltas[i] = delta_full

        # Update main path weights, biases, and contribution scores
        for i in range(1, self.num_layers):
            prev_active_mask = self.active_neurons.get(i-1, np.ones(self.weights[i].shape[0], dtype=bool))
            curr_active_mask = self.active_neurons[i]

            d_weights = np.dot(activations[i-1][:, prev_active_mask].T, deltas[i][:, curr_active_mask])

            # --- Contribution Tracking ---
            # Add a small decay to all scores
            self.neuron_contributions[i] *= 0.999
            # Add the current impact to the neurons that were just updated
            if d_weights.shape[1] > 0: # Ensure there are active neurons to update
                neuron_impact = np.mean(np.abs(d_weights), axis=0)
                self.neuron_contributions[i][curr_active_mask] += neuron_impact

            # Update weights and biases
            update_matrix = np.zeros_like(self.weights[i])
            ix_ = np.ix_(prev_active_mask, curr_active_mask)
            update_matrix[ix_] = d_weights
            self.weights[i] -= self.learning_rate * update_matrix
            self.biases[i][:, curr_active_mask] -= self.learning_rate * np.sum(deltas[i][:, curr_active_mask], axis=0)

    def _prune_network(self, prune_threshold=0.0):
        """Removes neurons with a contribution score below the threshold."""
        print("\n--- Checking for Pruning ---")
        for i in range(1, self.num_layers - 1): # Only prune hidden layers
            for neuron_idx in range(self.layer_sizes[i]):
                # Check if neuron is active and its score is below threshold
                if self.active_neurons[i][neuron_idx] and self.neuron_contributions[i][neuron_idx] < prune_threshold:
                    self.active_neurons[i][neuron_idx] = False
                    print(f"Pruned Neuron {neuron_idx} in Layer {i}. Score: {self.neuron_contributions[i][neuron_idx]:.4f}")
        print("--------------------------\n")

    def predict(self, X):
        """Make predictions on new data with the trained network."""
        predictions = []
        for x_sample in X:
            x_sample = x_sample.reshape(1, -1)
            # Use a simplified forward pass for prediction that respects pruning
            activations = {0: x_sample}
            for i in range(1, self.num_layers - 1):
                prev_active_neurons = self.active_neurons.get(i-1, np.ones(self.weights[i].shape[0], dtype=bool))
                active_w = self.weights[i][np.ix_(prev_active_neurons, self.active_neurons[i])]
                active_b = self.biases[i][:, self.active_neurons[i]]

                z_full = np.zeros((1, self.layer_sizes[i]))
                z_active = np.dot(activations[i-1][:, prev_active_neurons], active_w) + active_b
                z_full[:, self.active_neurons[i]] = z_active
                activations[i] = relu(z_full)

            output_layer_idx = self.num_layers - 1
            prev_active_neurons = self.active_neurons[output_layer_idx - 1]
            z_out = np.dot(activations[output_layer_idx-1][:, prev_active_neurons], self.weights[output_layer_idx][prev_active_neurons, :]) + self.biases[output_layer_idx]
            pred = linear(z_out)

            shortcut_output = 0
            for (layer_idx, neuron_idx), weight in self.shortcuts.items():
                neuron_activation = activations[layer_idx][0, neuron_idx]
                shortcut_output += neuron_activation * weight

            final_pred = pred + shortcut_output
            predictions.append(final_pred.flatten()[0])
        return np.array(predictions)

# --- Main Execution ---
if __name__ == '__main__':
    # 1. Load the California Housing dataset
    housing = fetch_california_housing()
    X_raw, y_raw = housing.data, housing.target

    # 2. Split data into training and testing sets
    X_train_raw, X_test_raw, y_train_raw, y_test_raw = train_test_split(X_raw, y_raw, test_size=0.2, random_state=42)

    # 3. Normalize the data (crucial for training)
    X_mean, X_std = X_train_raw.mean(axis=0), X_train_raw.std(axis=0)
    y_mean, y_std = y_train_raw.mean(), y_train_raw.std()

    X_train = (X_train_raw - X_mean) / X_std
    y_train = (y_train_raw - y_mean) / y_std
    X_test = (X_test_raw - X_mean) / X_std
    y_test = (y_test_raw - y_mean) / y_std


    # 4. Define and train the network
    # Architecture: 8 inputs -> 16 neurons -> 10 neurons -> 1 output
    # We've slightly increased layer sizes for the more complex data.
    layer_config = [X_train.shape[1], 16, 10, 1]
    nn = DynamicPruningNN(layer_sizes=layer_config, learning_rate=0.0001) # Using a smaller learning rate

    epochs = 100 # Fewer epochs for this larger dataset
    loss_history = nn.train(X_train, y_train, epochs=epochs, prune_interval=25, shortcut_threshold=0.1)

    # 5. Plot the training loss
    plt.figure(figsize=(10, 6))
    plt.plot(range(epochs), loss_history)
    plt.title('Training Loss Over Time (California Housing)')
    plt.xlabel('Epoch')
    plt.ylabel('Mean Squared Error')
    plt.grid(True)
    plt.show()

    # 6. Make predictions on the test set and compare
    predictions_normalized = nn.predict(X_test)
    # De-normalize predictions to see them in the original scale
    predictions = predictions_normalized * y_std + y_mean

    test_loss = mean_squared_error(y_test_raw, predictions)
    print(f"\nFinal Test Set MSE: {test_loss:.4f}")

    plt.figure(figsize=(10, 6))
    plt.scatter(y_test_raw, predictions, alpha=0.5, edgecolors='k')
    plt.plot([y_test_raw.min(), y_test_raw.max()], [y_test_raw.min(), y_test_raw.max()], 'r--', lw=2)
    plt.title('Actual vs. Predicted Housing Prices (Test Set)')
    plt.xlabel('Actual Price ($100k)')
    plt.ylabel('Predicted Price ($100k)')
    plt.grid(True)
    plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import fetch_california_housing
from sklearn.model_selection import train_test_split

# --- Activation Functions and Derivatives ---
# We use ReLU for hidden layers for its simplicity and efficiency,
# and a linear function for the output layer, which is standard for regression tasks.
def relu(x):
    return np.maximum(0, x)

def relu_derivative(x):
    return np.where(x > 0, 1, 0)

def linear(x):
    return x

def linear_derivative(x):
    return np.ones_like(x)

# --- Loss Function ---
# Mean Squared Error is a common choice for regression problems.
def mean_squared_error(y_true, y_pred):
    return np.mean((y_true - y_pred) ** 2)

def mse_derivative(y_true, y_pred):
    return y_pred - y_true

class DynamicPruningNN:
    """
    A neural network with a novel architecture featuring:
    1. Inline Backpropagation: Layer-by-layer updates during the forward pass.
    2. Dynamic Shortcuts: Neurons can bypass layers if they become predictive.
    3. Contribution-Based Pruning: Removes neurons with negative contributions.
    """
    def __init__(self, layer_sizes, learning_rate=0.001):
        self.layer_sizes = layer_sizes
        self.learning_rate = learning_rate
        self.num_layers = len(layer_sizes)

        # --- Initialize Network Parameters ---
        # We use dictionaries to store weights and biases for clarity.
        self.weights = {}
        self.biases = {}
        # We track active neurons for pruning. All start as active.
        self.active_neurons = {}
        # Contribution scores help decide which neurons to prune.
        self.neuron_contributions = {}
        # Stores the dynamic shortcut connections.
        self.shortcuts = {}

        for i in range(1, self.num_layers):
            # Initialize weights with small random values to break symmetry.
            self.weights[i] = np.random.randn(layer_sizes[i-1], layer_sizes[i]) * 0.1
            self.biases[i] = np.zeros((1, layer_sizes[i]))
            # All neurons are active initially.
            self.active_neurons[i] = np.ones(layer_sizes[i], dtype=bool)
            # Contributions start at zero.
            self.neuron_contributions[i] = np.zeros(layer_sizes[i])

    def _temp_predict(self, start_layer_idx, activation_in):
        """
        Performs a temporary forward pass from a given hidden layer to the output.
        This is the core of the "inline" feedback mechanism.
        """
        temp_activation = activation_in
        for i in range(start_layer_idx, self.num_layers - 1):
            # Get the active neuron masks for the previous and current layers in this temp pass
            prev_active_neurons = self.active_neurons.get(i, np.ones(self.weights[i+1].shape[0], dtype=bool))
            curr_active_neurons = self.active_neurons[i+1]

            # Get the sub-matrix of weights connecting only active neurons
            active_w = self.weights[i+1][np.ix_(prev_active_neurons, curr_active_neurons)]
            active_b = self.biases[i+1][:, curr_active_neurons]

            # Filter the input activation from the previous step to match the weight matrix
            filtered_input = temp_activation[:, prev_active_neurons]

            # Calculate the 'z' value for only the active part of the network
            z_active = np.dot(filtered_input, active_w) + active_b

            # Create a full-sized 'z' vector (padded with zeros) to maintain consistent shapes
            z_full = np.zeros((1, self.layer_sizes[i+1]))
            z_full[:, curr_active_neurons] = z_active

            if i < self.num_layers - 2: # Hidden layers
                temp_activation = relu(z_full)
            else: # Output layer
                temp_activation = linear(z_full)
        return temp_activation

    def train(self, X, y, epochs, prune_interval=50, shortcut_threshold=0.1):
        """
        Main training loop for the network.
        """
        loss_history = []
        print("--- Starting Training ---")
        for epoch in range(epochs):
            total_loss = 0
            # Shuffle data each epoch
            permutation = np.random.permutation(X.shape[0])
            X_shuffled = X[permutation]
            y_shuffled = y[permutation]

            for x_sample, y_sample in zip(X_shuffled, y_shuffled):
                x_sample = x_sample.reshape(1, -1)
                y_sample = y_sample.reshape(1, -1)

                # --- Forward Pass and Dynamic Shortcut Creation ---
                final_pred, activations, zs = self._forward_pass_and_check_shortcuts(x_sample, y_sample, shortcut_threshold, epoch)

                total_loss += mean_squared_error(y_sample, final_pred)

                # --- Unified Backpropagation ---
                self._backpropagate(y_sample, final_pred, activations, zs)

            # --- Periodic Pruning ---
            if (epoch + 1) % prune_interval == 0:
                self._prune_network()

            avg_loss = total_loss / len(X)
            loss_history.append(avg_loss)
            if (epoch + 1) % 10 == 0:
                print(f"Epoch {epoch+1}/{epochs}, Loss: {avg_loss:.6f}")

        print("--- Training Finished ---")
        return loss_history

    def _forward_pass_and_check_shortcuts(self, x_sample, y_sample, shortcut_threshold, epoch):
        """
        Performs a full forward pass while also checking if any neurons
        should be converted into shortcuts.
        """
        activations = {0: x_sample}
        zs = {}
        # Hidden layers
        for i in range(1, self.num_layers - 1):
            prev_active_neurons = self.active_neurons.get(i-1, np.ones(self.weights[i].shape[0], dtype=bool))
            active_w = self.weights[i][np.ix_(prev_active_neurons, self.active_neurons[i])]
            active_b = self.biases[i][:, self.active_neurons[i]]

            z_full = np.zeros((1, self.layer_sizes[i]))
            z_active = np.dot(activations[i-1][:, prev_active_neurons], active_w) + active_b
            z_full[:, self.active_neurons[i]] = z_active

            zs[i] = z_full
            activations[i] = relu(zs[i])

            # --- Inline Check for Shortcut Creation ---
            temp_pred = self._temp_predict(i, activations[i])
            local_error = mse_derivative(y_sample, temp_pred)
            if np.mean(np.abs(local_error)) < shortcut_threshold:
                best_neuron_idx = np.argmax(activations[i])
                if self.active_neurons[i][best_neuron_idx] and (i, best_neuron_idx) not in self.shortcuts:
                    shortcut_weight = np.random.randn(1,1) * 0.1
                    self.shortcuts[(i, best_neuron_idx)] = shortcut_weight
                    # print(f"Epoch {epoch}: Created shortcut from Layer {i}, Neuron {best_neuron_idx}") # Optional: uncomment for verbose logging

        # Output layer
        output_layer_idx = self.num_layers - 1
        prev_active_neurons = self.active_neurons[output_layer_idx - 1]

        z_out = np.dot(activations[output_layer_idx-1][:, prev_active_neurons], self.weights[output_layer_idx][prev_active_neurons, :]) + self.biases[output_layer_idx]
        zs[output_layer_idx] = z_out
        activations[output_layer_idx] = linear(zs[output_layer_idx])

        # Add shortcut contributions to the final output
        shortcut_output = 0
        for (layer_idx, neuron_idx), weight in self.shortcuts.items():
            neuron_activation = activations[layer_idx][0, neuron_idx]
            shortcut_output += neuron_activation * weight

        final_output = activations[output_layer_idx] + shortcut_output
        return final_output, activations, zs

    def _backpropagate(self, y, y_pred, activations, zs):
        """
        Performs a unified backpropagation pass that updates all main path weights,
        shortcut weights, and contribution scores.
        """
        deltas = {}

        # Start with the error at the output layer
        output_idx = self.num_layers - 1
        error_signal = mse_derivative(y, y_pred)
        deltas[output_idx] = error_signal * linear_derivative(zs[output_idx])

        # --- Update Shortcut Weights ---
        for (layer_idx, neuron_idx), weight in self.shortcuts.items():
            neuron_activation = activations[layer_idx][0, neuron_idx]
            # Gradient for shortcut is error * neuron's activation
            d_shortcut_weight = error_signal * neuron_activation
            self.shortcuts[(layer_idx, neuron_idx)] -= self.learning_rate * d_shortcut_weight

        # Propagate error backwards through hidden layers
        for i in range(output_idx - 1, 0, -1):
            error = np.dot(deltas[i+1], self.weights[i+1].T)
            delta_full = np.zeros((1, self.layer_sizes[i]))
            active_error = error[:, self.active_neurons[i]]
            active_z = zs[i][:, self.active_neurons[i]]
            delta_active = active_error * relu_derivative(active_z)
            delta_full[:, self.active_neurons[i]] = delta_active
            deltas[i] = delta_full

        # Update main path weights, biases, and contribution scores
        for i in range(1, self.num_layers):
            prev_active_mask = self.active_neurons.get(i-1, np.ones(self.weights[i].shape[0], dtype=bool))
            curr_active_mask = self.active_neurons[i]

            d_weights = np.dot(activations[i-1][:, prev_active_mask].T, deltas[i][:, curr_active_mask])

            # --- Contribution Tracking ---
            # Add a small decay to all scores
            self.neuron_contributions[i] *= 0.999
            # Add the current impact to the neurons that were just updated
            if d_weights.shape[1] > 0: # Ensure there are active neurons to update
                neuron_impact = np.mean(np.abs(d_weights), axis=0)
                self.neuron_contributions[i][curr_active_mask] += neuron_impact

            # Update weights and biases
            update_matrix = np.zeros_like(self.weights[i])
            ix_ = np.ix_(prev_active_mask, curr_active_mask)
            update_matrix[ix_] = d_weights
            self.weights[i] -= self.learning_rate * update_matrix
            self.biases[i][:, curr_active_mask] -= self.learning_rate * np.sum(deltas[i][:, curr_active_mask], axis=0)

    def _prune_network(self, prune_threshold=0.0):
        """Removes neurons with a contribution score below the threshold."""
        print("\n--- Checking for Pruning ---")
        for i in range(1, self.num_layers - 1): # Only prune hidden layers
            for neuron_idx in range(self.layer_sizes[i]):
                # Check if neuron is active and its score is below threshold
                if self.active_neurons[i][neuron_idx] and self.neuron_contributions[i][neuron_idx] < prune_threshold:
                    self.active_neurons[i][neuron_idx] = False
                    print(f"Pruned Neuron {neuron_idx} in Layer {i}. Score: {self.neuron_contributions[i][neuron_idx]:.4f}")
        print("--------------------------\n")

    def predict(self, X):
        """Make predictions on new data with the trained network."""
        predictions = []
        for x_sample in X:
            x_sample = x_sample.reshape(1, -1)
            # Use a simplified forward pass for prediction that respects pruning
            activations = {0: x_sample}
            for i in range(1, self.num_layers - 1):
                prev_active_neurons = self.active_neurons.get(i-1, np.ones(self.weights[i].shape[0], dtype=bool))
                active_w = self.weights[i][np.ix_(prev_active_neurons, self.active_neurons[i])]
                active_b = self.biases[i][:, self.active_neurons[i]]

                z_full = np.zeros((1, self.layer_sizes[i]))
                z_active = np.dot(activations[i-1][:, prev_active_neurons], active_w) + active_b
                z_full[:, self.active_neurons[i]] = z_active
                activations[i] = relu(z_full)

            output_layer_idx = self.num_layers - 1
            prev_active_neurons = self.active_neurons[output_layer_idx - 1]
            z_out = np.dot(activations[output_layer_idx-1][:, prev_active_neurons], self.weights[output_layer_idx][prev_active_neurons, :]) + self.biases[output_layer_idx]
            pred = linear(z_out)

            shortcut_output = 0
            for (layer_idx, neuron_idx), weight in self.shortcuts.items():
                neuron_activation = activations[layer_idx][0, neuron_idx]
                shortcut_output += neuron_activation * weight

            final_pred = pred + shortcut_output
            predictions.append(final_pred.flatten()[0])
        return np.array(predictions)

# --- Main Execution ---
if __name__ == '__main__':
    # 1. Load the California Housing dataset
    housing = fetch_california_housing()
    X_raw, y_raw = housing.data, housing.target

    # 2. Split data into training and testing sets
    X_train_raw, X_test_raw, y_train_raw, y_test_raw = train_test_split(X_raw, y_raw, test_size=0.2, random_state=42)

    # 3. Normalize the data (crucial for training)
    X_mean, X_std = X_train_raw.mean(axis=0), X_train_raw.std(axis=0)
    y_mean, y_std = y_train_raw.mean(), y_train_raw.std()

    X_train = (X_train_raw - X_mean) / X_std
    y_train = (y_train_raw - y_mean) / y_std
    X_test = (X_test_raw - X_mean) / X_std
    y_test = (y_test_raw - y_mean) / y_std


    # 4. Define and train the network
    # Architecture: 8 inputs -> 16 neurons -> 10 neurons -> 1 output
    # We've slightly increased layer sizes for the more complex data.
    layer_config = [X_train.shape[1], 16, 10, 1]
    nn = DynamicPruningNN(layer_sizes=layer_config, learning_rate=0.0001) # Using a smaller learning rate

    epochs = 100 # Fewer epochs for this larger dataset
    loss_history = nn.train(X_train, y_train, epochs=epochs, prune_interval=25, shortcut_threshold=0.1)

    # 5. Plot the training loss
    plt.figure(figsize=(10, 6))
    plt.plot(range(epochs), loss_history)
    plt.title('Training Loss Over Time (California Housing)')
    plt.xlabel('Epoch')
    plt.ylabel('Mean Squared Error')
    plt.grid(True)
    plt.show()

    # 6. Make predictions on the test set and compare
    predictions_normalized = nn.predict(X_test)
    # De-normalize predictions to see them in the original scale
    predictions = predictions_normalized * y_std + y_mean

    test_loss = mean_squared_error(y_test_raw, predictions)
    print(f"\nFinal Test Set MSE: {test_loss:.4f}")

    plt.figure(figsize=(10, 6))
    plt.scatter(y_test_raw, predictions, alpha=0.5, edgecolors='k')
    plt.plot([y_test_raw.min(), y_test_raw.max()], [y_test_raw.min(), y_test_raw.max()], 'r--', lw=2)
    plt.title('Actual vs. Predicted Housing Prices (Test Set)')
    plt.xlabel('Actual Price ($100k)')
    plt.ylabel('Predicted Price ($100k)')
    plt.grid(True)
    plt.show()
```
